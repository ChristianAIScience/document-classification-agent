```python

def extract_pdf_content(pdf_path: str) -> list[dict]:
    """
    Extrahiert Text und Tabellen aus einer PDF-Datei.
    """

    pdf_path = Path(pdf_path)

    if not pdf_path.exists():
        raise FileNotFoundError(
            f"PDF-Datei wurde nicht gefunden: {pdf_path}"
        )

    pages = []

    with pdfplumber.open(pdf_path) as pdf:

        for page_number, page in enumerate(
            pdf.pages,
            start=1,
        ):

            # --------------------------------------------------
            # Text extrahieren
            # --------------------------------------------------

            text = page.extract_text() or ""

            # --------------------------------------------------
            # Tabellen extrahieren
            # --------------------------------------------------

            raw_tables = page.extract_tables()

            tables = []

            for table in raw_tables:

                cleaned_table = []

                for row in table:

                    cleaned_row = [
                        cell.strip()
                        if isinstance(cell, str)
                        else cell
                        for cell in row
                    ]

                    cleaned_table.append(cleaned_row)

                tables.append(cleaned_table)

            # --------------------------------------------------
            # Seite speichern
            # --------------------------------------------------

            pages.append({
                "page": page_number,
                "text": text.strip(),
                "tables": tables,
            })

    return pages


if __name__ == "__main__":

    PDF_PATH = r"C:\Programming\multi-agent-system\data\uploads\pdf\firma\test_rech.pdf"
    

    try:
        content = extract_pdf_content(PDF_PATH)

        for page in content:

            print("=" * 60)
            print(f"SEITE {page['page']}")
            print("=" * 60)

            print("\nTEXT:")
            print(page["text"])

            print("\nTABELLEN:")

            for table_number, table in enumerate(
                page["tables"],
                start=1,
            ):

                print(f"\nTabelle {table_number}:")

                for row in table:
                    print(row)

    except Exception as error:
        print(f"❌ Fehler bei der PDF-Extraktion: {error}")
  
```
