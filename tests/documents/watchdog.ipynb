```python
import time

from pathlib import Path
from watchdog.observers import Observer
from watchdog.events import FileSystemEventHandler
```

```python

class PDFHandler(FileSystemEventHandler):
    def on_created(self, event):
        # Prüfen, ob es sich um eine Datei (keinen Ordner) handelt
        if event.is_directory:
            return
        print("dfsdgosd")
        file_path = Path(event.src_path)
        
        # Nur auf PDF-Dateien reagieren
        if file_path.suffix.lower() == ".pdf":
            print(f"\n[INFO] Neue PDF erkannt: {file_path.name}")
            
            # WICHTIG: Kurz warten, falls die Datei noch vom Kopieren/Download blockiert ist
            time.sleep(1) 
            
            try:
                self.process_pdf(file_path)
            except Exception as e:
                print(f"[FEHLER] Fehler bei der Verarbeitung von {file_path.name}: {e}")

    def process_pdf(self, pdf_path: Path):
        print(1)
        print(f"-> Starte OCR / Verarbeitung für: {pdf_path.name}")
        
        # HIER DEINE LOGIK EINBAUEN:
        # 1. PDF einlesen (z.B. mit deiner PaddleOCR / Extraction-Logik)
        # 2. In Markdown umwandeln oder direkt parsen
        # 3. In SQLModel (SQL) speichern
        # 4. Den Agenten triggern, der das Angebot mit der Rechnung vergleicht!
        
        # Beispiel: Eine einfache Dummy-Markdown-Datei als Ergebnis erzeugen
        output_md = pdf_path.with_suffix(".md")
        output_md.write_text(f"# Automatisch generierter Bericht für {pdf_path.name}\n\nStatus: Erfolgreich verarbeitet.")
        print(f"-> Markdown erstellt: {output_md.name}")
        
        # Optional: Original-PDF verschieben, damit sie nicht doppelt verarbeitet wird
        # processed_dir = Path("processed_pdfs")
        # processed_dir.mkdir(exist_ok=True)
        # pdf_path.rename(processed_dir / pdf_path.name)


# Überwachungs-Ordner definieren
WATCH_DIR = Path(r"C:\Programming\multi-agent-system\data\uploads\pdf")
WATCH_DIR.mkdir(exist_ok=True)


if __name__ == "__main__":
    event_handler = PDFHandler()
    observer = Observer()
    
    # Ordner im Hintergrund überwachen
    observer.schedule(event_handler, path=str(WATCH_DIR), recursive=False)
    observer.start()
    print(WATCH_DIR)
    print(f"Überwache Ordner: ./{WATCH_DIR.name}/ ... (Drücke Strg+C zum Beenden)")
    
    try:
        while True:
            time.sleep(1)
    except KeyboardInterrupt:
        observer.stop()
        print("\n[INFO] Watcher gestoppt.")
        
    observer.join()
```
